```python
import logging

# Configure logging
logging.getLogger('cmdstanpy').setLevel(logging.WARNING)
logging.getLogger('prophet').setLevel(logging.WARNING)

def forecast_tire_pressure(df, position):
    try:
        # Filter and prepare data
        position_df = df[df['Tire Position'] == position]
        position_df = position_df.sort_values('Timestamp').reset_index(drop=True)
        
        # Prophet requirements
        prophet_df = position_df[['Timestamp', 'Pressure (PSI)']].rename(
            columns={'Timestamp': 'ds', 'Pressure (PSI)': 'y'}
        )
        
        # Configure model with conservative settings
        model = Prophet(
            daily_seasonality=True,
            weekly_seasonality=False,  # Disable weekly seasonality
            yearly_seasonality=False,
            changepoint_prior_scale=0.05,  # More conservative than default
            seasonality_prior_scale=10.0,
            mcmc_samples=0
        )
        
        model.fit(prophet_df)
        
        # Create future dataframe (48 hours forecast)
        future = model.make_future_dataframe(periods=48, freq='H')
        forecast = model.predict(future)
        
        return forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']]
    
    except Exception as e:
        print(f"Error forecasting {position}: {str(e)}")
        return None

# Load data
df = pd.read_csv('tire_pressure_UAE_region.csv')
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Process each tire position
positions = df['Tire Position'].unique()
for position in positions:
    print(f"Processing {position}...")
    forecast = forecast_tire_pressure(df, position)
    if forecast is not None:
        print(f"Forecast for {position}:")
        print(forecast.tail())
```
